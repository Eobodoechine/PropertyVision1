Natural conversation with Gemini for property research

```python
import os
import json
from google.auth import default
import vertexai
from vertexai.generative_models import GenerativeModel

# For Vertex AI Workbench - use default credentials
try:
    credentials, project_id = default()
    print(f"✅ Using project: {project_id}")
    
    # Initialize Vertex AI
    vertexai.init(project=project_id, location='us-central1')
    print("✅ Vertex AI initialized")
    
except Exception as e:
    print(f"❌ Authentication failed: {e}")
    print("Make sure you're running in Vertex AI Workbench with proper permissions")
```

```python
# Initialize the Gemini model
model = GenerativeModel('gemini-1.5-pro')

# Test connection
try:
    response = model.generate_content("Hello! Can you help me research real estate data?")
    print("✅ Gemini connected successfully!")
    print(f"Response: {response.text}")
except Exception as e:
    print(f"❌ Connection failed: {e}")
```

```python
# Natural chat function
def chat(message):
    """Chat naturally with Gemini"""
    try:
        response = model.generate_content(message)
        print(f"🤖 Gemini: {response.text}")
        return response.text
    except Exception as e:
        print(f"❌ Error: {e}")
        return None

# Example usage
chat("I need help investigating real estate data discrepancies. Can you help me research property sale dates?")
```

## Advanced: Chat with Search Tools
This version includes Google Search capabilities

```python
# Chat with search tools in Vertex AI
def chat_with_search(message):
    """Chat with Google Search grounding enabled"""
    try:
        # Use Vertex AI model with search tools
        model_with_search = GenerativeModel(
            'gemini-1.5-pro',
            tools=[{"google_search": {}}]  # Enable search grounding
        )
        
        response = model_with_search.generate_content(message)
        print(f"🔍 Gemini (with search): {response.text}")
        return response.text
    except Exception as e:
        print(f"❌ Search chat error: {e}")
        print("Falling back to regular chat...")
        return chat(message)

# Test search capabilities
chat_with_search("Search for 105 Bailey Ct Fayetteville GA 30215 recent sale information")
```
